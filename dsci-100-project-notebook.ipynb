#### Task 5.2 Perform the Analysis (20 points)

```r
# Set the seed to make the analysis reproducible
set.seed(51151892)

# Split the data into 75% training and 25% testing
basketball_split <- initial_split(basketball_tidy,
                                  prop = 0.75)

basketball_train <- training(basketball_split)
basketball_test <- testing(basketball_split)
```

```r
# Select the response and predictor variables
# Standardize the predictors so they have similar scales
basketball_recipe <- recipe(
    tourney_wins ~ reg_wins_pct + conf_rank,
    data = basketball_train
) |>
    step_scale(all_predictors()) |>
    step_center(all_predictors())

basketball_recipe
```

```r
# Create a KNN regression model and tune the number of neighbours
basketball_spec <- nearest_neighbor(
    weight_func = "rectangular",
    neighbors = tune()
) |>
    set_engine("kknn") |>
    set_mode("regression")

basketball_spec
```

```r
# Perform five-fold cross-validation
basketball_vfold <- vfold_cv(basketball_train, v = 5)

# Combine the recipe and model into a workflow
basketball_workflow <- workflow() |>
    add_recipe(basketball_recipe) |>
    add_model(basketball_spec)

basketball_workflow
```

```r
# Create a range of possible K values
gridvals <- tibble(neighbors = seq(from = 1,
                                   to = 100,
                                   by = 1))

# Calculate cross-validation RMSPE for each K value
basketball_results <- basketball_workflow |>
    tune_grid(resamples = basketball_vfold,
              grid = gridvals) |>
    collect_metrics() |>
    filter(.metric == "rmse")

basketball_results
```

```r
# Find the K value with the lowest cross-validation RMSPE
basketball_min <- basketball_results |>
    arrange(mean) |>
    head(1)

basketball_min

# Extract the best number of neighbours
basketball_k <- basketball_min |>
    pull(neighbors)

basketball_k
```

```r
# Create the final model using the best K value
basketball_best_spec <- nearest_neighbor(
    weight_func = "rectangular",
    neighbors = basketball_k
) |>
    set_engine("kknn") |>
    set_mode("regression")

# Fit the final model using the entire training dataset
basketball_fit <- workflow() |>
    add_recipe(basketball_recipe) |>
    add_model(basketball_best_spec) |>
    fit(data = basketball_train)

basketball_fit
```

```r
# Predict tournament wins using the testing dataset
basketball_predictions <- basketball_fit |>
    predict(basketball_test) |>
    bind_cols(basketball_test)

head(basketball_predictions)
```

```r
# Calculate the RMSPE using the testing dataset
basketball_metrics <- basketball_predictions |>
    metrics(truth = tourney_wins,
            estimate = .pred) |>
    filter(.metric == "rmse")

basketball_metrics
```

```r
# Calculate the average tournament wins in the training data
baseline_mean <- basketball_train |>
    summarize(mean_wins = mean(tourney_wins)) |>
    pull(mean_wins)

# Calculate the RMSPE when predicting the training mean
baseline_metrics <- basketball_test |>
    mutate(baseline_prediction = baseline_mean) |>
    metrics(truth = tourney_wins,
            estimate = baseline_prediction) |>
    filter(.metric == "rmse")

baseline_metrics
```
